*Figura 3.1: Diagramas de blocos de uma malha digital com PID Digital 1.*

Definindo $T(q^{-1}) = R(q^{-1})$, o PID Digital 1 enquadra-se na estrutura $RST$. A função de transferência em malha fechada entre a referência $r(t)$ e a saída $y(t)$ é:

$$H_{CL}(q^{-1}) = \frac{B(q^{-1}) R(q^{-1})}{A(q^{-1})S(q^{-1}) + B(q^{-1})R(q^{-1})} = \frac{B(q^{-1})R(q^{-1})}{P(q^{-1})} $$

Onde:
R/S definimoos como controlador e B/A é definida como a planta;



Controlador digital PID 1:

Projeção: 
- Obtenção do modelo discreto da planta;
- Especificações de desempenho em malha fechada;
- Cálculo dos coeficientes do controlador (R e S);
- Verificação das margens de robustez e das funções de sensibilidade.

1. Modelo Planta discreta:

Para o caso de um sistema identificado a planta já será digital/discreta.

Adotando uma planta de segunda ordem na forma:

$$ H(q^{-1}) = \frac{b_1 q^{-1} + b_2 q^{-2}} {1 + a_1 q^{-1} + a_2 q^{-2}} = \frac{B(q^{-1})}{A(q^{-1})} $$

2. Especificações de desempenho em malha fechada:

De acordo com a equação de malha fechada do controlador pid digital 1, é obtido:

$$ H_{CL}(q^{-1}) = \frac{B(q^{-1})\,R(q^{-1})} {P(q^{-1})} $$

Obs: Onde $P(q^{-1})$  define os polos em malha fechada desejados.

- Grau:

$Gr(A') = Gr(A) + Gr(H(q^{-1})) = 3;
\\
Gr(B) = 2$;

Portanto:

$$Gr(R(q^{-1})) = Gr(A') - 1 = 2
\\
Gr(S'(q^{-1})) = Gr(B) + d - 1 = 1
\\
\therefore Gr(S(q^{-1})) = Gr(H(q^{-1})) + Gr(S'(q^{-1})) = 2$$


para o polinômio característico:

$Gr(P(q^{-1})) = Gr(A') + Gr(B) - 1 = 4$

$$\therefore P(q^{-1}) =  (1 + p'_1 q^{-1} + p'_2 q^{-2})(1 + \beta_1 q^{-1})(1 + \beta_2 q^{-2}) = 1 + p_1 q^{-1} + p_2 q^{-2} + p_3 q^{-3} + p_4 q^{-4} $$

onde $\beta_1$ e $\beta_2$ são **polos auxiliares aperiódicos**, situados no eixo real dentro do círculo unitário, escolhidos tipicamente no intervalo:

$$-0{,}05 \le \beta_{1,2} \le -0{,}5 $$

3. Cálculo dos coeficientes do controlador (R e S):

Para determinar $R(q^{-1})$ e $S(q^{-1})$, resolve-se a identidade polinomial de Bézout:

$$A'(q^{-1}) S(q^{-1}) + B(q^{-1}) R(q^{-1}) = P(q^{-1})



Igualando as potências de $q^{-k}$, obtém-se o sistema linear:

$$M x = p $$

onde:
$$x = \begin{bmatrix} s'_1 & r_0 & r_1 & r_2 \end{bmatrix}^T $$

$$p = \begin{bmatrix} p_1 - a'_1 & p_2 - a'_2 & p_3 - a'_3 & p_4 \end{bmatrix}^T $$

sendo $A'(q^{-1}) = A(q^{-1})(1 - q^{-1}) = 1 + a'_1 q^{-1} + a'_2 q^{-2} + a'_3 q^{-3}$.

A matriz de Sylvester associada $M$ possui dimensão $4 \times 4$. A solução única:

$$x = M^{-1} p $$

Obs.:
termo integrador = $H(q^{-1}) = (1 - q^{-1})$

4. Recuperação dos Parâmetros Contínuos Equivalentes
Caso se deseje avaliar os ganhos contínuos correspondentes:

$$K = -\frac{r_1 + 2 r_2}{1 + s'_1}$$

$$T_i = T_s \frac{K(1 + s'_1)}{r_0 + r_1 + r_2} $$

$$T_d = T_s \frac{s'_1 r_1 - (1 - s'_1) r_2}{K (1 + s'_1)^3} $$

$$\frac{T_d}{N} = -T_s \frac{s'_1}{1 + s'_1} $$

Para haver equivalência física com um PID contínuo com filtro derivativo, é estritamente necessário que:

$$-1 < s'_1 < 0 $$

Se $0 < s'_1 < 1$, o controlador digital continua estável e funcional em malha fechada, porém **não possui correspondente no domínio contínuo**.

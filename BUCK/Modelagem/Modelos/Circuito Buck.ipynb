#### $$BuckModelo$$

Modelagem matemática.

O Conversor buck pode ser analisado em dois estados principais, são eles:

1. Estado 1:

Em primeiro momento entende-se que a chave presente no circuito estaja fechada, alimentando o circuito com a tensão da fonte.

2. Estado 2:

Em contraponto, o segundo momento a chave estará aberta, fazendo com que o nivel de tensão seja determinado a partir do capacitor. 


Portanto, a modelagem matemática é realizado para cada tipo de operação, neste caso, duas operações, chave on ou chave off.


---

#### **1° estado:** chave fechada

[  ] Por LKT:

$$\sum_{i=1}^{n} V_i = 0$$

portanto:

$$Vf = Vl + Vc = \frac{diL}{dt}\cdot L + Vc$$

$$\frac{diL}{dt}\cdot L = Vf - Vc = \frac{diL}{dt} = \frac{Vf}{L} - \frac{Vc}{L}$$

eq. final (1):

$$\frac{diL}{dt} = \frac{Vf}{L} - \frac{Vc}{L}$$







[  ] Por LKC:

$$\sum_{j=1}^{n} i_j = 0$$

Portanto: 

$$iL = iC + ir = iL = \frac{dvC}{dt} \cdot L + ir$$


eq. final (2):


$$\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{R\cdot C}$$


- A partir das equações (1) e (2):


$$\begin{cases}
\frac{diL}{dt} = \frac{Vf}{L} - \frac{Vc}{L} \\
\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{R\cdot C}
\end{cases}$$

---

#### **2° estado:** chave aberta

---

[  ] Por LKT:

$$\sum_{i=1}^{n} V_i = 0$$

portanto:

$$vL = -vC$$

eq. final (3):

$$\frac{diL}{dt} = \frac{VC}{L}$$


[  ] Por LKC:

$$\sum_{j=1}^{n} i_j = 0$$

Portanto: 

$$iL = iC + ir = \frac{dvC}{dt}\cdot C + \frac{Vcarga}{R}$$

eq. final (4):


$$\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{C}$$

- A partir das equações (3) e (4):


$$\begin{cases}
\frac{diL}{dt} = \frac{VC}{L} \\
\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{C}
\end{cases}$$

---





---
- Simbolos:

Vf = Tensão na fonte 
Vc = Tensão capacitor
Vl = Tensão indutor
L = Indultancia
$ \frac{diL}{dt} = $ taxa de variação de corrente em relação ao tempo no indutor
$ \frac{dvC}{dt} = $ taxa de variação de tensão em relação ao tempo no capacito
R = Resistencia carga 
C = Capacitância 
if = corrente fonte 
iL = corrente indutor 
ir = corrente carga
i = corrente
iC = corrente capacitor
Vcarga = tensão carga

 $$BuckModelo$$

Modelagem no espaço de estados.

A partir da modelagem matemática obtem-se o modelo no espaço de estados.

---

1. Definir variáveis de estado.
2. Definir o modelo na equação de estados.
3. Definição do modelo médio.

---


## 1° Definição das variaveis de estado

As principais variaveis que determinam o estado dinamico do conversor buck são:

- variação de corrente no indutor.
- variação de tensão no capacitor.


---


## 2° Definir o modelo na equação de estados.


**Estado (1) chave fechada**


obtem-se


para x1

$$x1 = iL(t)$$

logo:

$$x1' = \frac{diL}{dt} = \frac{U1}{L} - \frac{x2}{L}$$

para x2:


$$x2 = Vc(t) $$

logo:

$$x2' = \frac{dvC}{dt} = \frac{x1}{C} - \frac{x2}{R\cdot C} $$


Portanto:

$$\begin{cases}
x1' = \frac{U1}{L} - \frac{x2}{L} \\
x2' = \frac{x1}{C} - \frac{x2}{R\cdot C}
\end{cases}$$

#### **Equação de estado estado(1):** 

Definido por:

$$X' = A1\cdot x + B1U$$

$$

\begin{bmatrix}
x1' \\
x2'
\end{bmatrix}
=
\begin{bmatrix}
0 & -\frac{1}{L} \\
\frac{1}{C} & -\frac{1}{RC}
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
+
\begin{bmatrix}
\frac{1}{L} \\
0
\end{bmatrix}
U1(t)

$$


---

### **Estado (2) chave aberta**

$$\begin{cases}
x1' = \frac{-x2}{L}  \\
x2' = \frac{x1}{C} - \frac{x2}{R\cdot C}
\end{cases}$$

#### **Equação de estado estado(2):** 

Definido por:

$$X' = A2\cdot x + B2U$$


$$

\begin{bmatrix}
x1' \\
x2'
\end{bmatrix}
=
\begin{bmatrix}
0 & -\frac{1}{L} \\
\frac{1}{C} & -\frac{1}{RC}
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
+
\begin{bmatrix}
0 \\
0
\end{bmatrix}
U1(t)

$$

## 3. Definição do modelo médio. 


Definição:

$$ M = k\cdot M1 + (1-k)\cdot M2$$


Portanto, para definição do modelo médio:

$$ M1 = k\cdot A1 + (1-k)\cdot A2$$


$$ M2 = k\cdot B1 + (1-k)\cdot B2$$


Obtem-se:

$$

M1 = \begin{bmatrix}
0 & \frac{-1}{L} \\
\frac{1}{C} & \frac{-1}{RC}
\end{bmatrix}
$$

$$
M2 = \begin{bmatrix}
\frac{-K}{L} \\
0
\end{bmatrix}
$$



O modelo médio é obtido por:

$$ Mmédio = k\cdot M1 + (1-k)\cdot M2$$


Portanto:


$$

Mmédio = \begin{bmatrix}
x1' \\
x2'
\end{bmatrix}
=
\begin{bmatrix}
0 & -\frac{1}{L} \\
\frac{1}{C} & -\frac{1}{RC}
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
+
\begin{bmatrix}
\frac{K}{L} \\
0
\end{bmatrix}
U1(t)

$$

## 1. Introdução
Este documento tem como objetivo apresentar e demonstrar a modelagem matemática para o conversor buck.

---

## 2. Fundamentação teórica.

O Conversor buck pode ser analisado em dois estados principais, são eles:

1. Estado 1:

Em primeiro momento entende-se que a chave presente no circuito estaja fechada, alimentando o circuito com a tensão da fonte.

2. Estado 2:

Em contraponto, o segundo momento a chave estará aberta, fazendo com que o nivel de tensão seja determinado a partir do capacitor. 


Portanto, a modelagem matemática é realizada para cada tipo de operação, neste caso, duas operações, chave on ou chave off.


---


## 3. Resultados e discursões



#### **1° estado:** chave fechada

$$\begin{cases}
\frac{diL}{dt} = \frac{Vf}{L} - \frac{Vc}{L} \\
\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{R\cdot C}
\end{cases}$$

#### **2° estado:** chave aberta


$$\begin{cases}
\frac{diL}{dt} = \frac{VC}{L} \\
\frac{dvC}{dt} = \frac{iL}{C} - \frac{Vcarga}{C}
\end{cases}$$

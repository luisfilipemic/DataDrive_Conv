A Escolha do modelo:

O modelo ARX (do inglês AutoRegressive with eXogenous input — AutoRegressivo com entrada eXógena) é considerado a estrutura fundamental e o "arquétipo" para a identificação de sistemas lineares. Ele descreve a relação entre o sinal de entrada u(k) e a saída y(k) de um sistema através de uma equação de diferenças lineares.

Estrutura:

- Equações de diferenças ARX, pode ser definida por:

$$ y(k) = \sum_{i=1}^{n} aiy(k-i) + \sum_{j=1}^{n} bjy(k-i) $$



## 3. Resultados e discussão:



Modelo obtido atráves do conjunto de dados:



- Equações de diferenças ARX, pode ser definida por:

$$ y(k) = \sum_{i=1}^{n} aiy(k-i) + \sum_{j=1}^{n} bjy(k-i) $$



Para dois coeficientes de entrada e saída, teremos:

$$y(k) = a1\cdot y(k-1) + a2\cdot y(k-2) + b1\cdot u(k-1) + b2\cdot u(k-2)$$

- Matriz dos regressores:

Temos que, K>3;

$$y(3) = a1\cdot y(2) + a2\cdot y(1) + b1\cdot u(2) + b2\cdot u(1)$$

$$y(4) = a1\cdot y(3) + a2\cdot y(2) + b1\cdot u(3) + b2\cdot u(2)$$

$$...$$

$$y(k) = a1\cdot y(k-1) + a2\cdot y(k-2) + b1\cdot u(k-1) + b2\cdot u(k-2)$$



Por definição, a partir do sistema linear podemos gerar a seguinte matriz:


$$A\cdot X = Y$$

$$
\begin{equation}
\underbrace{
\begin{bmatrix} 
y(3) \\ 
y(4) \\ 
\vdots \\ 
y(N) 
\end{bmatrix}
}_{Y}
=
\underbrace{
\begin{bmatrix} 
y(2) & y(1) & u(2) & u(1) \\
y(3) & y(2) & u(3) & u(2) \\
\vdots & \vdots & \vdots & \vdots \\
y(N-1) & y(N-2) & u(N-1) & u(N-2)
\end{bmatrix}
}_{A}
\underbrace{
\begin{bmatrix} 
a_1 \\ 
a_2 \\ 
b_1 \\ 
b_2 
\end{bmatrix}
}_{X}
\end{equation}
$$


Em que A é denotada como a matriz dos regressores




Isolando X atráves dos minimos quadrados, teremos:

$$


(A\cdot X = y)


$$

$$

X = (A^T \cdot A)^{-1}\cdot A^T\cdot Y
$$


Com a o algoritimo dos minimos quadrados executado, basta encontrar os coeficientes da matriz X.
